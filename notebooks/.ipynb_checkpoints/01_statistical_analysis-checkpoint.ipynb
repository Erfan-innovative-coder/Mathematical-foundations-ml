```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Define a small dataset for statistical analysis
data = np.array([2, 4, 4, 4, 5, 7])
```

```python
# Calculate descriptive statistics
mean_value = np.mean(data)
variance_value = np.var(data, ddof=1)
std_value = np.std(data, ddof=1)
median_value = np.median(data)
q1, q3 = np.percentile(data, [25, 75])

# Summarize the results in a DataFrame
summary = pd.DataFrame({
    "Metric": [
        "Mean",
        "Median",
        "Sample Variance",
        "Sample Standard Deviation",
        "First Quartile",
        "Third Quartile"
    ],
    "Value": [
        mean_value,
        median_value,
        variance_value,
        std_value,
        q1,
        q3
    ]
})

summary
```

```python
# Visualize the data distribution and central tendency
plt.figure(figsize=(8, 5))
plt.hist(data, bins="auto", edgecolor="black")
plt.axvline(mean_value, linestyle="--", label="Mean")
plt.axvline(median_value, linestyle=":", label="Median")

plt.title("Distribution and Descriptive Statistics")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()

# Save the figure for the project report
plt.savefig(
    "../reports/figures/descriptive_statistics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Calculate additional measures of data dispersion
data_range = np.max(data) - np.min(data)
iqr = q3 - q1

dispersion = pd.DataFrame({
    "Metric": ["Range", "Interquartile Range"],
    "Value": [data_range, iqr]
})

dispersion
```
